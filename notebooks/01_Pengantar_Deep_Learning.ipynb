## 3. Konsep Dasar: Neuron Biologis vs Neuron Buatan

### Neuron Biologis
- **Dendrit**: Menerima sinyal input
- **Soma (cell body)**: Memproses sinyal
- **Akson**: Mengirim output ke neuron lain
- **Sinapsis**: Koneksi antar neuron

### Neuron Buatan (Perceptron)
- **Input (x)**: Data yang diterima
- **Bobot (w)**: Kekuatan koneksi
- **Bias (b)**: Nilai penyesuaian
- **Fungsi Aktivasi**: Menentukan apakah neuron 'aktif' atau tidak
- **Output (y)**: Hasil komputasi

**Formula:**
$$y = f(\sum_{i} w_i x_i + b)$$

````python
# Install dan import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

print("Library berhasil diimport!")
print(f"NumPy version: {np.__version__}")
````

## 4. Implementasi Neuron Tunggal (Perceptron)

````python
class Neuron:
    """
    Implementasi sederhana sebuah neuron buatan.
    """
    def __init__(self, n_inputs):
        # Inisialisasi bobot secara acak
        self.weights = np.random.randn(n_inputs)
        self.bias = np.random.randn()
        
    def sigmoid(self, z):
        """Fungsi aktivasi Sigmoid: mengubah nilai ke rentang (0, 1)"""
        return 1 / (1 + np.exp(-z))
    
    def forward(self, inputs):
        """Hitung output neuron"""
        # Weighted sum
        z = np.dot(self.weights, inputs) + self.bias
        # Fungsi aktivasi
        output = self.sigmoid(z)
        return output

# Buat neuron dengan 3 input
neuron = Neuron(n_inputs=3)

# Contoh input
inputs = np.array([1.0, 0.5, -1.0])

# Hitung output
output = neuron.forward(inputs)

print("=" * 40)
print("Simulasi Neuron Tunggal")
print("=" * 40)
print(f"Input    : {inputs}")
print(f"Bobot    : {neuron.weights.round(4)}")
print(f"Bias     : {neuron.bias:.4f}")
print(f"Output   : {output:.4f}")
````

## 5. Fungsi Aktivasi

Fungsi aktivasi menentukan apakah dan seberapa besar sinyal yang diteruskan oleh neuron. Tanpa fungsi aktivasi, neural network hanya bisa memodelkan hubungan linear.

````python
def sigmoid(z):
    """Sigmoid: output antara 0 dan 1"""
    return 1 / (1 + np.exp(-z))

def tanh(z):
    """Tanh: output antara -1 dan 1"""
    return np.tanh(z)

def relu(z):
    """ReLU: Rectified Linear Unit - yang paling populer"""
    return np.maximum(0, z)

def leaky_relu(z, alpha=0.01):
    """Leaky ReLU: mengatasi masalah dying ReLU"""
    return np.where(z > 0, z, alpha * z)

# Visualisasi
z = np.linspace(-5, 5, 200)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Fungsi-Fungsi Aktivasi dalam Deep Learning', fontsize=14, fontweight='bold')

# Plot Sigmoid
axes[0, 0].plot(z, sigmoid(z), color='blue', linewidth=2)
axes[0, 0].set_title('Sigmoid')
axes[0, 0].set_xlabel('z')
axes[0, 0].set_ylabel('σ(z)')
axes[0, 0].axhline(y=0, color='k', linestyle='-', linewidth=0.5)
axes[0, 0].axvline(x=0, color='k', linestyle='-', linewidth=0.5)
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].set_ylim(-0.1, 1.1)
axes[0, 0].annotate('σ(z) = 1/(1+e⁻ᶻ)', xy=(1, 0.7), fontsize=9, color='blue')

# Plot Tanh
axes[0, 1].plot(z, tanh(z), color='green', linewidth=2)
axes[0, 1].set_title('Tanh')
axes[0, 1].set_xlabel('z')
axes[0, 1].set_ylabel('tanh(z)')
axes[0, 1].axhline(y=0, color='k', linestyle='-', linewidth=0.5)
axes[0, 1].axvline(x=0, color='k', linestyle='-', linewidth=0.5)
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].set_ylim(-1.1, 1.1)

# Plot ReLU
axes[1, 0].plot(z, relu(z), color='red', linewidth=2)
axes[1, 0].set_title('ReLU (Rectified Linear Unit)')
axes[1, 0].set_xlabel('z')
axes[1, 0].set_ylabel('ReLU(z)')
axes[1, 0].axhline(y=0, color='k', linestyle='-', linewidth=0.5)
axes[1, 0].axvline(x=0, color='k', linestyle='-', linewidth=0.5)
axes[1, 0].grid(True, alpha=0.3)

# Plot Leaky ReLU
axes[1, 1].plot(z, leaky_relu(z), color='purple', linewidth=2)
axes[1, 1].set_title('Leaky ReLU')
axes[1, 1].set_xlabel('z')
axes[1, 1].set_ylabel('Leaky ReLU(z)')
axes[1, 1].axhline(y=0, color='k', linestyle='-', linewidth=0.5)
axes[1, 1].axvline(x=0, color='k', linestyle='-', linewidth=0.5)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 6. Arsitektur Neural Network

Neural Network terdiri dari:
- **Input Layer**: Menerima data masukan
- **Hidden Layer(s)**: Lapisan pemrosesan tersembunyi
- **Output Layer**: Menghasilkan prediksi

Deep Learning = Neural Network dengan **banyak** hidden layer

````python
def visualize_network(layer_sizes, title="Neural Network Architecture"):
    """
    Visualisasi arsitektur neural network.
    layer_sizes: list jumlah neuron per layer, contoh [3, 4, 4, 2]
    """
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.set_title(title, fontsize=14, fontweight='bold', pad=15)
    
    n_layers = len(layer_sizes)
    max_neurons = max(layer_sizes)
    
    layer_colors = ['#2196F3', '#4CAF50', '#4CAF50', '#F44336']
    if n_layers > 4:
        layer_colors = ['#2196F3'] + ['#4CAF50'] * (n_layers - 2) + ['#F44336']
    
    neuron_positions = []
    
    for layer_idx, n_neurons in enumerate(layer_sizes):
        x = layer_idx / (n_layers - 1)
        positions = []
        for neuron_idx in range(n_neurons):
            y = (neuron_idx + 1) / (n_neurons + 1)
            positions.append((x, y))
            color = layer_colors[min(layer_idx, len(layer_colors)-1)]
            circle = plt.Circle((x, y), 0.025, color=color, zorder=4, ec='white', lw=1.5)
            ax.add_patch(circle)
        neuron_positions.append(positions)
    
    # Gambar koneksi
    for layer_idx in range(n_layers - 1):
        for pos1 in neuron_positions[layer_idx]:
            for pos2 in neuron_positions[layer_idx + 1]:
                ax.plot([pos1[0], pos2[0]], [pos1[1], pos2[1]],
                        'gray', alpha=0.3, linewidth=0.5, zorder=1)
    
    # Label layer
    layer_labels = ['Input\nLayer'] + [f'Hidden\nLayer {i+1}' for i in range(n_layers-2)] + ['Output\nLayer']
    for layer_idx, label in enumerate(layer_labels):
        x = layer_idx / (n_layers - 1)
        ax.text(x, -0.05, label, ha='center', va='top', fontsize=9,
                fontweight='bold', color=layer_colors[min(layer_idx, len(layer_colors)-1)])
        ax.text(x, 1.05, f'{layer_sizes[layer_idx]} neuron', ha='center', va='bottom',
                fontsize=8, color='gray')
    
    # Legend
    legend_elements = [
        mpatches.Patch(color='#2196F3', label='Input Layer'),
        mpatches.Patch(color='#4CAF50', label='Hidden Layer'),
        mpatches.Patch(color='#F44336', label='Output Layer')
    ]
    ax.legend(handles=legend_elements, loc='upper right', fontsize=9)
    
    ax.set_xlim(-0.1, 1.1)
    ax.set_ylim(-0.1, 1.15)
    ax.axis('off')
    plt.tight_layout()
    plt.show()

# Visualisasi neural network sederhana
print("Contoh 1: Neural Network dengan 1 hidden layer")
visualize_network([3, 4, 2], "Neural Network Sederhana (3-4-2)")
````

````python
print("Contoh 2: Deep Neural Network dengan 3 hidden layer")
visualize_network([4, 6, 6, 4, 2], "Deep Neural Network (4-6-6-4-2)")
````

## 7. Implementasi Neural Network dari Scratch

Mari kita implementasikan neural network sederhana untuk klasifikasi menggunakan hanya NumPy.

````python
class SimpleNeuralNetwork:
    """
    Neural Network sederhana dengan 1 hidden layer.
    Arsitektur: Input -> Hidden -> Output
    """
    
    def __init__(self, n_input, n_hidden, n_output, learning_rate=0.1):
        self.lr = learning_rate
        
        # Inisialisasi bobot dengan nilai kecil acak
        self.W1 = np.random.randn(n_input, n_hidden) * 0.01
        self.b1 = np.zeros((1, n_hidden))
        self.W2 = np.random.randn(n_hidden, n_output) * 0.01
        self.b2 = np.zeros((1, n_output))
    
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-np.clip(z, -500, 500)))
    
    def sigmoid_derivative(self, z):
        s = self.sigmoid(z)
        return s * (1 - s)
    
    def forward(self, X):
        """Forward pass: hitung prediksi"""
        # Layer 1
        self.Z1 = np.dot(X, self.W1) + self.b1
        self.A1 = self.sigmoid(self.Z1)
        
        # Layer 2 (output)
        self.Z2 = np.dot(self.A1, self.W2) + self.b2
        self.A2 = self.sigmoid(self.Z2)
        
        return self.A2
    
    def compute_loss(self, y_true, y_pred):
        """Binary Cross-Entropy Loss"""
        m = y_true.shape[0]
        epsilon = 1e-15
        loss = -np.mean(y_true * np.log(y_pred + epsilon) + 
                        (1 - y_true) * np.log(1 - y_pred + epsilon))
        return loss
    
    def backward(self, X, y_true):
        """Backward pass: hitung gradien"""
        m = X.shape[0]
        
        # Gradien output layer
        dA2 = self.A2 - y_true
        dZ2 = dA2 * self.sigmoid_derivative(self.Z2)
        dW2 = np.dot(self.A1.T, dZ2) / m
        db2 = np.sum(dZ2, axis=0, keepdims=True) / m
        
        # Gradien hidden layer
        dA1 = np.dot(dZ2, self.W2.T)
        dZ1 = dA1 * self.sigmoid_derivative(self.Z1)
        dW1 = np.dot(X.T, dZ1) / m
        db1 = np.sum(dZ1, axis=0, keepdims=True) / m
        
        # Update bobot (Gradient Descent)
        self.W2 -= self.lr * dW2
        self.b2 -= self.lr * db2
        self.W1 -= self.lr * dW1
        self.b1 -= self.lr * db1
    
    def train(self, X, y, epochs=1000, verbose=True):
        """Latih model"""
        losses = []
        
        for epoch in range(epochs):
            # Forward pass
            y_pred = self.forward(X)
            
            # Hitung loss
            loss = self.compute_loss(y, y_pred)
            losses.append(loss)
            
            # Backward pass
            self.backward(X, y)
            
            if verbose and epoch % 200 == 0:
                accuracy = np.mean((y_pred > 0.5).astype(int) == y)
                print(f"Epoch {epoch:4d} | Loss: {loss:.4f} | Accuracy: {accuracy:.4f}")
        
        return losses

print("Kelas SimpleNeuralNetwork berhasil dibuat!")
````

````python
# Buat dataset XOR - masalah klasik yang tidak bisa diselesaikan perceptron tunggal
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])

y = np.array([[0],
              [1],
              [1],
              [0]])

print("Dataset XOR:")
print("Input  | Output")
print("-" * 20)
for i in range(len(X)):
    print(f"{X[i]}  |  {y[i][0]}")
````

````python
# Latih neural network
np.random.seed(42)
nn = SimpleNeuralNetwork(n_input=2, n_hidden=4, n_output=1, learning_rate=1.0)

print("Training Neural Network untuk masalah XOR...")
print("=" * 55)
losses = nn.train(X, y, epochs=1000, verbose=True)

print("\nHasil Prediksi Akhir:")
print("=" * 35)
predictions = nn.forward(X)
for i in range(len(X)):
    pred = predictions[i][0]
    label = 1 if pred > 0.5 else 0
    print(f"Input: {X[i]} | Prediksi: {pred:.4f} | Label: {label} | Benar: {y[i][0]} | {'✓' if label == y[i][0] else '✗'}")
````

````python
# Visualisasi proses training
plt.figure(figsize=(10, 4))
plt.plot(losses, color='blue', linewidth=1.5)
plt.title('Kurva Learning: Loss vs Epoch', fontsize=13, fontweight='bold')
plt.xlabel('Epoch')
plt.ylabel('Loss (Binary Cross-Entropy)')
plt.grid(True, alpha=0.3)
plt.fill_between(range(len(losses)), losses, alpha=0.1, color='blue')
plt.tight_layout()
plt.show()

print(f"\nLoss awal : {losses[0]:.4f}")
print(f"Loss akhir: {losses[-1]:.4f}")
print(f"Penurunan : {((losses[0] - losses[-1]) / losses[0] * 100):.1f}%")
````
